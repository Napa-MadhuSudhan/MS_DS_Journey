# Five Number Summary
### This is used to Identify the outliers from the data

*   Minimum
*   Q1 value
*   Medium
*   Q3 Value
*   Maximum

```
Data distribution = [lower fence, upper Fence]
Lower Fence = Q1- 1.5(IQR)
Upper Fence = Q3 + 1.5(IQR)
IQR( Inter Quartail Range) =  Q3-Q1
Q3 = (75/100)*(N+1)
Q1 = (25/100)*(N+1)
N = Totoal population
```





## Lets Solve it using some example of tips data

````python
total_bill = df['total_bill']
````

````python
N = len(total_bill)
````

````python
Q3 = np.quantile(total_bill, 0.75)
````

````python
Q1 = np.quantile(total_bill, 0.25)
````

````python
Lower_fence = Q1 - (1.5*(Q3-Q1))
````

````python
Upper_fence = Q3 + (1.5*(Q3-Q1))
````

````python
Data_distribution = [Lower_fence, Upper_fence]
````

````python
print(Data_distribution)
````

````python
total_bill = total_bill.sort_values()
````

````python
lower_outliers = total_bill[total_bill < Data_distribution[0]]
````

````python
upper_outliers = total_bill[total_bill > Data_distribution[1]]
````

````python
# Lets see outliers clearly with the help of boxplot
sns.boxplot(total_bill)
````

We can see a lot outliers in the upper fence .

lets drop all the outliers

````python
total_bill = total_bill.drop(lower_outliers.index)
````

````python
total_bill = total_bill.drop(upper_outliers.index)
````

````python
sns.boxplot(total_bill)
````

## Using Five number summary we are able to remove outliers from the data . final boxplot can visualise the all the data is well distrubted
